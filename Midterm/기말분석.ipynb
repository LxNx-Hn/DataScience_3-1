# final.csv — Gold 분포 확인

아래 셀은 `Midterm/final.csv`에서 `gold` 컬럼을 읽어와서
- 기초 통계(`describe`) 출력
- 히스토그램 + KDE
- 박스플롯
를 순서대로 보여줍니다. 셀을 실행하면 분포를 시각적으로 확인할 수 있습니다.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# 파일 경로 (노트북이 repo 루트에서 열려 있다고 가정)
path = 'Midterm/final.csv'

# 데이터 읽기
try:
    df = pd.read_csv(path)
except Exception as e:
    raise SystemExit(f"파일을 읽는 중 오류가 발생했습니다: {e}")

print('파일에서 읽은 컬럼:', list(df.columns))

col = 'gold'
if col not in df.columns:
    print(f"경고: '{col}' 컬럼이 없습니다. 사용 가능한 컬럼: {list(df.columns)}")
else:
    s = df[col].dropna()
    print('\n기초 통계량:')
    print(s.describe())

    # 히스토그램 + KDE
    plt.figure(figsize=(10,5))
    sns.histplot(s, bins=30, kde=True, color='C0')
    plt.title('Gold 분포 — 히스토그램 + KDE')
    plt.xlabel('gold')
    plt.ylabel('count')
    plt.tight_layout()
    plt.show()

    # 박스플롯
    plt.figure(figsize=(6,4))
    sns.boxplot(x=s, color='C1')
    plt.title('Gold 분포 — 박스플롯')
    plt.xlabel('gold')
    plt.tight_layout()
    plt.show()

    # 추가: 분포가 매우 치우쳐있다면 로그 스케일 예시
    if (s > 0).sum() > 0 and (s.max() / (s.replace(0, pd.NA).median() or 1)) > 100:
        plt.figure(figsize=(10,5))
        sns.histplot(s[s>0].apply(lambda x: np.log1p(x)), bins=30, kde=True, color='C2')
        plt.title('Gold 분포 (log1p 변환)')
        plt.xlabel('log1p(gold)')
        plt.tight_layout()
        plt.show()
```
